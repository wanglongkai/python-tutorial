```python
status_code = 200
if status_code == 200:
    print('请求成功')
elif status_code == 404:
    print('资源不存在')
elif status_code == 500:
    print('服务器内部错误')
elif status_code >= 400:
    print('请求失败')
else:
    print('未知状态码')
```

```python
status_code = 200
match status_code:
    case 200:
        print("请求成功")
    case 404:
        print("资源不存在")
    case _:
        print("其他状态码")
```
