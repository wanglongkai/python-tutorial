```python
"""
海象运算符 :=    

声明变量并赋值
注意：海象运算符的赋值部分通常建议加括号，以避免优先级问题。
"""

# 传统写法： 先赋值，后使用
n= len('wanglongkai')
if n >5 :
    print(f'字符的长度是{n}')


# 海象运算发写法
if (m := len('wanglongkai')) > 5:
    print(f'海象运算：字符长度{m}')    
```

```python
# 传统写法
line = file.readline()
while line != "":
    print(line)
    line = file.readline()  # 重复代码

# 海象写法
while (line := file.readline()) != "":
    print(line)  # 简洁，无重复
```
