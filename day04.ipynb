# 高阶函数

函数作为参数参与逻辑

````python
def sum(a, b, f):
    return f(a) + f(b)

result = sum(1, 2, abs)
print(result)
````

## 内置高阶函数

````python
# map
list1 = [1, 2, 3, 4, 5]
list2 = [x ** 2 for x in list1]
print(list2)

list3 = list(map(lambda x: x ** 2, list1))
print(list3)
````

````python
import functools

list4 = [1,2,3,4,5]
list5 = functools.reduce(lambda acc,cur: acc+cur, list4)
print(list5)
````

````python
list6 = [1,2,3,4,5]

list7 = list(filter(lambda x: x % 2 ==0, list6))
print(list7)
````

# lambda表达式（匿名函数）

1. 参数可有可无，也可以有多个
2. 但必须有且仅有一个返回值

````python
# 无参数
lambdafn1 = lambda: 100
print(lambdafn1())
# 多个参数
lambdafn2 = lambda x, y ,z: x + y + z
print(lambdafn2(1, 2, 3))
# 有默认参数
lambdafn3 = lambda x, y ,z=100: x + y + z
print(lambdafn3(1, 2))
#可变参数
lambdafn4 = lambda *args: [x ** 2 for x in args]
print(lambdafn4(1, 2, 3, 4, 5))
# 关键字参数
lambdafn5 = lambda **kwargs: kwargs
print(lambdafn5(name='张三', age=18))
````
