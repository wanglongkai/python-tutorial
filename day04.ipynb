```python
with open('RAG/README.md', 'r', encoding='utf-8') as f:
    f.seek(0, 2) # 移动文件指针到文件末尾， seek(offset, whence) whence=0表示从文件开头开始，whence=1表示从当前位置开始，whence=2表示从文件末尾开始
    lines = f.readlines()
    for index, line in enumerate(lines):
        print(index, line)
```

```python
with open('RAG/README.md', 'r', encoding='utf-8') as f:
    with open('RAG/README[备份].md', 'w', encoding='utf-8') as f2:
        f2.write(f.read())
```
